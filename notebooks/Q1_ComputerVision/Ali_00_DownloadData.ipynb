```python
import os
from huggingface_hub import login, snapshot_download

# ورود به اکانت
login()

# تنظیم مسیر
RAW_DATA_DIR = r"D:\GitHub\QueraAiBootcamp-TeamProject2-ComputerVision&NLP\data\raw\nexar_collision_prediction"
os.makedirs(RAW_DATA_DIR, exist_ok=True)

print("در حال اتصال به سرور Hugging Face و شروع دانلود...")

# دانلود کل مخزن دیتاست
downloaded_path = snapshot_download(
    repo_id="nexar-ai/nexar_collision_prediction",
    repo_type="dataset",
    local_dir=RAW_DATA_DIR,
    local_dir_use_symlinks=False # دانلود فایل‌های واقعی به جای شورت‌کات
)

print("-" * 40)
print(f" دانلود با موفقیت کامل شد!")
print(f" فایل‌ها در این مسیر قرار دارند: {downloaded_path}")
```
