```python
train_df['vote'] = pd.to_numeric(train_df['vote'], errors='coerce').fillna(0)
```

```python
top_clients = train_df.groupby(["reviewerID", "reviewerName"])["vote"].sum().reset_index(name="total_vote")
top_clients
```

```python
top_10 = top_clients.sort_values("total_vote", ascending=False).head(10).reset_index(drop=True)
top_10.index += 1

top_10
```
