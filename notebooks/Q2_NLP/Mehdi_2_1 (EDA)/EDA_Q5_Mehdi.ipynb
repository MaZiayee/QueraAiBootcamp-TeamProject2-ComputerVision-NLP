```python
five_star = train_df[train_df["overall"] == 5]
five_star
```

```python
top_products = five_star.groupby("asin").size().reset_index(name="five_star_count")
top_products = top_products.merge(product_df[["asin", "brand", "title"]], on="asin", how="left")
top_10 = top_products.sort_values("five_star_count", ascending=False).head(10)
top_10 = top_10[["brand", "title", "five_star_count"]].reset_index(drop=True)
top_10.index += 1

top_10
```
