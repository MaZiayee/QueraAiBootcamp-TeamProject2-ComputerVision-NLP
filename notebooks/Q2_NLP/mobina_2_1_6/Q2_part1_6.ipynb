```python
# خواندن جدول نگاشت شناسه‌ی محصولات به عنوان و برند آن‌ها
product_map = pd.read_csv("product_mapping.csv")

# نمایش چند سطر اول
product_map.head()
```

```python
# بررسی نام ستون‌های قدیم
df.columns
```

```python
# اتصال برند به دیتاست اصلی بر اساس Asin
df = df.merge(product_map, on="asin", how="left")

# بررسی نام ستون‌های جدید
df.columns
```

```python
# پیدا کردن 10 برند با بیشترین تعداد نظر
top10_brands = (df["brand"].value_counts().head(10).index)

# محاسبه تعداد نظرات و میانگین امتیاز برای این برندها
brand_analysis = (df[df["brand"].isin(top10_brands)].groupby("brand")
    .agg(review_count=("overall", "count"), average_rating=("overall", "mean"))
    .reset_index())

# مرتب‌سازی بر اساس میانگین امتیاز
brand_analysis = brand_analysis.sort_values(by="average_rating", ascending=False)

# تصحیح ایندکس‌ها
brand_analysis = brand_analysis.reset_index(drop=True)

# شروع ایندکس‌ها از یک به‌جای صفر
brand_analysis.index = brand_analysis.index + 1

# نمایش جدول نهایی
brand_analysis
```
